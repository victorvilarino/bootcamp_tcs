```python
# LIMPEZA BASICA

import pandas as pd

df = pd.read_csv("iris_sujo.csv")

print(df.head())
print(df.info())

print(df.isnull().sum()) # Nulos por coluna
print(df.isna().sum(axis=1)) # Nulos por linha

df = df.dropna()

```

```python
# GRAFICO

data_plot = df.reset_index()

data_plot["Species"].hist()
```

```python
numeric_columns = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]

print(f"{df[numeric_columns].mean()}\n")

print(f"{df[numeric_columns].median()}\n")

print(f"{df[numeric_columns].max()}\n")

print(f"{df[numeric_columns].min()}\n")
```

```python
df_species_mean = df.groupby("Species")[numeric_columns].mean()

df_species_mean
```
