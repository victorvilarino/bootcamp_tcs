```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("dados.csv")

df = df.drop(columns='Carimbo de data/hora')

df.isnull().sum()
```

```python
it_field_percentage = df['Área de atuação que mais se adequa para sua realidade:'
                         ].value_counts(normalize=True) * 100

working_model_satisfaction = df.groupby('Modalidade de atuação que se aplica para sua realidade:'
                                        )['Nível de satisfação com o mercado de trabalho:'].mean()

print(it_field_percentage)
print()
print(working_model_satisfaction)
```

```python
plt.Figure(figsize=(10, 5))

sns.barplot(
    x=it_field_percentage.values,
    y=it_field_percentage.index
)

plt.title("Percentual por área de atuação")
plt.xlabel("Porcentagem (%)")
plt.ylabel("Área de atuação")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=working_model_satisfaction.index,
    y=working_model_satisfaction.values
)

plt.title("Satisfação média por modalidade de trabalho")
plt.xlabel("Modalidade")
plt.ylabel("Satisfação média")

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")

plt.show()
```
